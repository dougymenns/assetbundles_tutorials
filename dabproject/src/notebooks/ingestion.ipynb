```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog_name}.default.pipeline_audit (
    pipeline_id STRING,
    run_date DATE,
    status STRING,
    message STRING,
    load_timestamp TIMESTAMP
)
""")

spark.sql(f"""
INSERT INTO {catalog_name}.default.pipeline_audit VALUES
    ('pipeline_01', DATE('2026-08-01'), 'SUCCESS', 'Pipeline executed successfully', CURRENT_TIMESTAMP()),
    ('pipeline_02', DATE('2026-08-02'), 'FAILED', 'Error during data load', CURRENT_TIMESTAMP()),
    ('pipeline_03', DATE('2026-08-03'), 'SUCCESS', 'All steps completed', CURRENT_TIMESTAMP())
""")

df_audit = spark.sql(f"SELECT * FROM {catalog_name}.default.pipeline_audit")
display(df_audit)
```
